```python
import pandas as pd
import numpy as np
```

```python
## read ntf data
ntf_df = pd.read_csv("data/NTF_deidentified_demographics_filled.csv")

## read tts data
tts_df = pd.read_csv("data/TTS_deidentified_demographics_filled.csv")

## combine dataframes
combined_df = pd.concat([ntf_df, tts_df], ignore_index=True)

## save combined dataframe
combined_df.to_csv("data/combined_demographics.csv", index=False)
```

```python
## read symptom scores
ntf_symptoms_df = pd.read_csv("data/NTFGH_symptom_score.csv")
tts_symptoms_df = pd.read_csv("data/TTSH_symptom_score.csv")

## combine symptom scores
combined_symptoms_df = pd.concat([ntf_symptoms_df, tts_symptoms_df], ignore_index=True)
## save combined symptom scores
combined_symptoms_df.to_csv("data/combined_symptom_scores.csv", index=False)
```
